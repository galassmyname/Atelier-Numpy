# Atelier NumPy — Analyse de données de capteurs IoT


```python
import numpy as np
```

## Partie 1 : Listes Python vs tableaux NumPy

On dispose des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0 et 25.8

### 1) Liste Python

```python
temperatures_liste = [24.5, 25.1, 26.3, 27.0, 25.8]
print(temperatures_liste)
```

### 2) Tableau NumPy équivalent

```python
temperatures_array = np.array([24.5, 25.1, 26.3, 27.0, 25.8])
print(temperatures_array)
```

### 3) Multiplication par 2

```python
# Sur la liste Python
liste_x2 = temperatures_liste * 2
print("Liste x2 :", liste_x2)

# Sur le tableau NumPy
array_x2 = temperatures_array * 2
print("Array x2 :", array_x2)
```

### 4) Différence observée

- Sur la **liste Python** (`temperatures_liste * 2`), l'opérateur `*` **duplique** la liste : on obtient 10 éléments (les 5 valeurs répétées deux fois), et non les valeurs doublées.
- Sur le **tableau NumPy** (`temperatures_array * 2`), l'opérateur `*` effectue une **multiplication élément par élément** : chaque température est bien multipliée par 2, et le tableau garde 5 éléments.

**Conclusion :** NumPy est conçu pour le calcul vectorisé, tandis que les listes Python traitent les opérateurs comme des opérations de séquence. C'est pourquoi NumPy est bien plus adapté au traitement de données numériques comme nos mesures de capteurs.

## Partie 2 : Création des données

### 1) Tableau de 10 capteurs, températures initialement à zéro

```python
temperatures_capteurs = np.zeros(10)
print(temperatures_capteurs)
```

### 2) Tableau de 10 valeurs égales à 1

```python
valeurs_uns = np.ones(10)
print(valeurs_uns)
```

### 3) Tableau de 10 valeurs initialisées à 25

```python
valeurs_25 = np.full(10, 25)
print(valeurs_25)
```

### 4) Matrice identité de taille 5x5

```python
matrice_identite = np.eye(5)
print(matrice_identite)
```

### 5) Les 24 heures (0 à 23) de mesure d'une journée

```python
heures = np.arange(24)
print(heures)
```

### 6) 10 valeurs régulièrement réparties entre 20°C et 30°C

```python
valeurs_reparties = np.linspace(20, 30, 10)
print(valeurs_reparties)
```

### 7) 100 mesures de températures aléatoires, résultats reproductibles

```python
np.random.seed(50) 
temperatures_100 = np.random.uniform(15, 35, 100)
print(temperatures_100)
```
