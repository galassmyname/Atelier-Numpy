## Partie 1 : Listes Python vs tableaux NumPy

On dispose des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0 et 25.8

### 1) Liste Python

```python
temperatures_liste = [24.5, 25.1, 26.3, 27.0, 25.8]
print(temperatures_liste)
```

### 2) Tableau NumPy équivalent

```python
temperatures_array = np.array([24.5, 25.1, 26.3, 27.0, 25.8])
print(temperatures_array)
```

### 3) Multiplication par 2

```python
# Sur la liste Python
liste_x2 = temperatures_liste * 2
print("Liste x2 :", liste_x2)

# Sur le tableau NumPy
array_x2 = temperatures_array * 2
print("Array x2 :", array_x2)
```

### 4) Différence observée

- Sur la **liste Python** (`temperatures_liste * 2`), l'opérateur `*` **duplique** la liste : on obtient 10 éléments (les 5 valeurs répétées deux fois), et non les valeurs doublées.
- Sur le **tableau NumPy** (`temperatures_array * 2`), l'opérateur `*` effectue une **multiplication élément par élément** : chaque température est bien multipliée par 2, et le tableau garde 5 éléments.

**Conclusion :** NumPy est conçu pour le calcul vectorisé, tandis que les listes Python traitent les opérateurs comme des opérations de séquence. C'est pourquoi NumPy est bien plus adapté au traitement de données numériques comme nos mesures de capteurs.

## Partie 3 : Exploration des tableaux

### 1) Tableau aléatoire des 24 températures d'une journée

```python
np.random.seed(42)
temperatures_jour = np.random.uniform(15, 35, 24)
print(temperatures_jour)
```

### 2) Propriétés du tableau

```python
# a) Nombre de dimensions
print("Nombre de dimensions :", temperatures_jour.ndim)

# b) Nombre de valeurs
print("Nombre de valeurs :", temperatures_jour.size)

# c) Type des données
print("Type des données :", temperatures_jour.dtype)

# d) Forme du tableau
print("Forme du tableau :", temperatures_jour.shape)
```

**Interprétation :**
- `ndim` = 1 → le tableau est unidimensionnel (un vecteur).
- `size` = 24 → il contient 24 valeurs (une par heure).
- `dtype` = float64 → les valeurs sont des nombres décimaux.
- `shape` = (24,) → 24 éléments sur une seule dimension.

### 1) Première mesure

```python
print("Première mesure (0h) :", temperatures_jour[0])
```

### 2) Dernière mesure

```python
print("Dernière mesure (23h) :", temperatures_jour[-1])
```

### 3) Température de 8h

```python
print("Température à 8h :", temperatures_jour[8])
```

### 4) Températures de 8h à 12h

```python
print("Températures de 8h à 12h :", temperatures_jour[8:13])
```

### 5) Températures des 6 premières heures

```python
print("6 premières heures :", temperatures_jour[:6])
```

### 6) Températures des heures paires

```python
print("Heures paires :", temperatures_jour[::2])
```

## Partie 5 : Filtrage booléen

On considère toujours le tableau `temperatures_jour`. On souhaite détecter les situations inhabituelles.

### 1) Températures supérieures à 30°C

```python
temperatures_sup_30 = temperatures_jour[temperatures_jour > 30]
print("Températures > 30°C :", temperatures_sup_30)
```

### 2) Températures « normales » (entre 20°C et 30°C)

```python
temperatures_normales = temperatures_jour[(temperatures_jour >= 20) & (temperatures_jour <= 30)]
print("Températures normales (20-30°C) :", temperatures_normales)
```

### 3) Tableau booléen des anomalies (> 35°C)

```python
anomalies = temperatures_jour > 35
print("Anomalies (booléen) :", anomalies)
```

### 4) Nombre de températures anormales

```python
nombre_anomalies = np.sum(anomalies)
print("Nombre de températures anormales :", nombre_anomalies)
```

## Partie 6 : Manipulation des dimensions

On suppose qu'on a 5 capteurs, chacun mesurant les 24 heures d'une journée.

### 1) Matrice aléatoire des températures des 5 capteurs sur une journée

```python
np.random.seed(42)
temperatures_5capteurs = np.random.uniform(15, 35, (5, 24))
print(temperatures_5capteurs)
```

### 2) Transformation en tableau unidimensionnel

```python
temperatures_flat = temperatures_5capteurs.flatten()
print(temperatures_flat)
print("Forme :", temperatures_flat.shape)
```

### 3) Transposée de la matrice

```python
temperatures_transposee = temperatures_5capteurs.T
print(temperatures_transposee)
print("Forme :", temperatures_transposee.shape)
```

### 4) Mesures du capteur 1

```python
capteur_1 = temperatures_5capteurs[0]
print(capteur_1)
```

### 5) Mesures du capteur 5

```python
capteur_5 = temperatures_5capteurs[4]
print(capteur_5)
```

### 6) Mesures de tous les capteurs à 10h

```python
mesures_10h = temperatures_5capteurs[:, 10]
print(mesures_10h)
```

### 7) Mesures des capteurs 2 à 4 entre 8h et 12h

```python
mesures_capteurs2a4_8a12h = temperatures_5capteurs[1:4, 8:13]
print(mesures_capteurs2a4_8a12h)
```

## Partie 7 : Concaténation

On suppose qu'on a un 2ème bâtiment avec 3 capteurs.

### 1) Matrice aléatoire des températures des 3 capteurs du 2ème bâtiment

```python
np.random.seed(24)
temperatures_batiment2 = np.random.uniform(15, 35, (3, 24))
print(temperatures_batiment2)
```

### 2) Fusion des températures des 8 capteurs (5 + 3), deux méthodes

```python
# Méthode 1 : np.concatenate
temperatures_8capteurs_v1 = np.concatenate((temperatures_5capteurs, temperatures_batiment2), axis=0)
print("Méthode 1 - forme :", temperatures_8capteurs_v1.shape)

# Méthode 2 : np.vstack
temperatures_8capteurs_v2 = np.vstack((temperatures_5capteurs, temperatures_batiment2))
print("Méthode 2 - forme :", temperatures_8capteurs_v2.shape)
```

**Note :** les deux méthodes empilent les matrices verticalement (ligne par ligne), donnant une matrice finale de forme (8, 24) — 8 capteurs, 24 heures. `np.concatenate` avec `axis=0` et `np.vstack` sont équivalents ici.
