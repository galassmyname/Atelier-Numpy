## Partie 1 : Listes Python vs tableaux NumPy

On dispose des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0 et 25.8

### 1) Liste Python

```python
temperatures_liste = [24.5, 25.1, 26.3, 27.0, 25.8]
print(temperatures_liste)
```

### 2) Tableau NumPy équivalent

```python
temperatures_array = np.array([24.5, 25.1, 26.3, 27.0, 25.8])
print(temperatures_array)
```

### 3) Multiplication par 2

```python
# Sur la liste Python
liste_x2 = temperatures_liste * 2
print("Liste x2 :", liste_x2)

# Sur le tableau NumPy
array_x2 = temperatures_array * 2
print("Array x2 :", array_x2)
```

### 4) Différence observée

- Sur la **liste Python** (`temperatures_liste * 2`), l'opérateur `*` **duplique** la liste : on obtient 10 éléments (les 5 valeurs répétées deux fois), et non les valeurs doublées.
- Sur le **tableau NumPy** (`temperatures_array * 2`), l'opérateur `*` effectue une **multiplication élément par élément** : chaque température est bien multipliée par 2, et le tableau garde 5 éléments.

**Conclusion :** NumPy est conçu pour le calcul vectorisé, tandis que les listes Python traitent les opérateurs comme des opérations de séquence. C'est pourquoi NumPy est bien plus adapté au traitement de données numériques comme nos mesures de capteurs.

## Partie 3 : Exploration des tableaux

### 1) Tableau aléatoire des 24 températures d'une journée

```python
np.random.seed(42)
temperatures_jour = np.random.uniform(15, 35, 24)
print(temperatures_jour)
```

### 2) Propriétés du tableau

```python
# a) Nombre de dimensions
print("Nombre de dimensions :", temperatures_jour.ndim)

# b) Nombre de valeurs
print("Nombre de valeurs :", temperatures_jour.size)

# c) Type des données
print("Type des données :", temperatures_jour.dtype)

# d) Forme du tableau
print("Forme du tableau :", temperatures_jour.shape)
```

**Interprétation :**
- `ndim` = 1 → le tableau est unidimensionnel (un vecteur).
- `size` = 24 → il contient 24 valeurs (une par heure).
- `dtype` = float64 → les valeurs sont des nombres décimaux.
- `shape` = (24,) → 24 éléments sur une seule dimension.

### 1) Première mesure

```python
print("Première mesure (0h) :", temperatures_jour[0])
```

### 2) Dernière mesure

```python
print("Dernière mesure (23h) :", temperatures_jour[-1])
```

### 3) Température de 8h

```python
print("Température à 8h :", temperatures_jour[8])
```

### 4) Températures de 8h à 12h

```python
print("Températures de 8h à 12h :", temperatures_jour[8:13])
```

### 5) Températures des 6 premières heures

```python
print("6 premières heures :", temperatures_jour[:6])
```

### 6) Températures des heures paires

```python
print("Heures paires :", temperatures_jour[::2])
```
